```python
plt.figure(figsize=(8, 5))

plt.hist(df["popularity"], bins=20, edgecolor="black")

plt.title("Распределение популярности треков Spotify")
plt.xlabel("Популярность трека")
plt.ylabel("Количество треков")

plt.tight_layout()
plt.show()
```

```python
df["track_genre"].value_counts().head(10).plot(
    kind="bar",
    edgecolor="black",
    figsize=(10, 5)
)

plt.title("Топ-10 музыкальных жанров Spotify")
plt.xlabel("Музыкальный жанр")
plt.ylabel("Количество треков")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```
